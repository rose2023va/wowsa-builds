# WOWSA Distance Runner

**To run the Ireland calculation:** Click **Runtime → Run all** in the menu above. Wait ~2 minutes. A zip file will download to your computer with the map, the Word doc, and the route data.

**To run a different swim:** Edit the parameters in cell 4 (mode, name, bbox, waypoints filename), then Run all again.

Files must be present in your Google Drive root (My Drive):
- `wowsa_distance.py`
- `waypoints_ireland.json` (or your swim's waypoint file)
- `labels_ireland.json` (optional — for map labels)

```
# Step 1: Connect to Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```
# Step 3: Copy files from Drive
import shutil
shutil.copy('/content/drive/MyDrive/wowsa_distance.py', '.')
shutil.copy('/content/drive/MyDrive/waypoints_ireland.json', '.')
shutil.copy('/content/drive/MyDrive/labels_ireland.json', '.')
print('Files copied.')
```

```
# Step 4: Run the calculation. Edit these parameters for other swims.
MODE = 'circumnavigation'
NAME = 'Ireland'
BBOX = '50.9,-11.0,55.8,-5.0'
WAYPOINTS = 'waypoints_ireland.json'
LABELS = 'labels_ireland.json'
OUTPUT = 'out_ireland'

!python wowsa_distance.py --mode {MODE} --name "{NAME}" --bbox {BBOX} --waypoints {WAYPOINTS} --labels {LABELS} --output {OUTPUT}
```

```
# Step 5: Zip the output folder and download it
!zip -r {OUTPUT}.zip {OUTPUT}
from google.colab import files
files.download(f'{OUTPUT}.zip')
```
